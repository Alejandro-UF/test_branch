```python
#Agrupamos por nacionalidad 
jugadores_pais = tabla.groupby('Nationality').size()

#Almacenamos en un dataset
ds = jugadores_pais.reset_index(name='Numero de jugadores')
ds
```

```python
#De igual forma sacamos el MIN,MEAN,MAX
print("Media de jugadores:", jugadores_pais.mean())
print("Mínimo de jugadores:", jugadores_pais.min())
print("Máximo de jugadores:", jugadores_pais.max())
```
